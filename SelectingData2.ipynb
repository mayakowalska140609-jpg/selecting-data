## create random samples

```python
df = pd.read_csv('players_20.csv')
df = df.set_index('short_name')
df = df[['long_name', 'age', 'height_cm', 'weight_kg', 'nationality', 'club']]
```

```python
# extract 10 random elements from nationality column
df['nationality'].sample(10, random_state=99)
```

```python
# extract a random 20% sample from the dataframe
df.sample(frac=0.2)
```

```python
# increasing the sampling rate --> frac>1
df.sample(frac=2, replace=True)
```

## filtering data using query() method

```python
df
```

```python
# select players older than 34
df.query("age>34")

#equivalent boolean slicing
df[df['age']>34]

# the same but use different syntax
```

```python
# select players older than 34 from italy
df.query("age>34 and nationality=='Italy'")

# boolean slicing equivalent
df[(df['age'] > 34) & (df['nationality'] == 'Italy')]
```

```python
# add a not operator to the first example
df.query("not (age>34)")

#boolean slicing
df[df['age']>~34]
```

```python
# convert height to m and select those with a height above 180
df.query("height_cm/100 > 1.8")

# boolean equivalent
df[df['height_cm']/100 > 1.8]
```

## apply()

```python
df = pd.read_csv('players_20.csv')
df = df.set_index('short_name')
df = df[['long_name', 'age', 'dob', 'height_cm', 'weight_kg', 'nationality', 'club']]
```

```python
# use numpy function and apply it to series
import numpy as np
df['age'].apply(np.sqrt)
```

```python
# create own function and apply to dataframe

def calculate_bmi(row):
    return row['weight_kg']/(row['height_cm'] / 100)**2
```

```python
df.apply(calculate_bmi, axis=1)
```

## lambda function + apply() method

```python
# basic function
def sum_values(a,b):
    x = a + b
    return x
```

```python
sum_values(2,3)
```

```python
sum_values
```

```python
# lambda function (one liner)
sum_values_lambda = lambda a,b:a+b
# colon separated input and output
```

```python
sum_values_lambda(2,3)
```

```python
# lambda to convert height_cm to metres
df['height_cm'].apply(lambda x:x/100)

#alternative
#df['height_cm']/100
```

```python
# lambda to convert long_name series to upper case
df['long_name'].apply(lambda x:x.upper())

#alternative with str attribute
df['long_name'].str.upper()
```

```python
df.dtypes
df['dob'] = df['dob'].astype('datetime64[ns]')
```

```python
# lambda to get year / dob series
#df['dob'].apply(lambda x:x.year)

#alternative with dt attribute
df['dob'].dt.year
```

```python
df
```

```python
# lambda to calculate bmi 



df.apply(lambda x:x['weight_kg'] / ((x['height_cm']/100)*2), axis=1)
```

## making a copy of the dataframe

```python
# deep = True by default
df_copy = df.copy()
```

```python
# update value in original dataframe
df.loc['L.Messi', 'height_cm'] = 180
```

```python
# copy vs og dataframe
df_copy
```

```python
# deep= False, any changes made to original will be reflected in the copy
df_copy = df.copy(deep=False)
```
