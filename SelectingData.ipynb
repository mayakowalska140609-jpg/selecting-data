```python
import pandas as pd
```

```python
df = pd.read_csv('players_20.csv)'
df
```

```python
# set column as index
df = df.set_index('short_name')
```

```python
df
```

```python
#select columns
df = df[['long_name', 'age', 'dob', 'height_cm', 'weight_kg', 'nationality', 'club']]
```

```python
df
```

## selecting a single value (loc method)

loc[row_label, column_label]

```python
# all data about messi
df.loc['L. Messi']
```

```python
# height of messi
df.loc['L. Messi', 'height_cm']
```

```python
# height of ronaldo
df.loc['Cristiano Ronaldo', 'height_cm']
```

```python
# get all rows inside height_cm column
df.loc[:, 'height_cm']
```

```python
# get all columns that correspong to the index L. Messi
df.loc['L. Messi',:]
```

## Selecting with a list of values

```python
# get all data about L. Messi and C. Ronaldo

df.loc[['L. Messi', 'Cristiano Ronaldo']]
```

```python
# get the height of L. Messi and C. Ronaldo
df.loc[['L. Messi', 'Cristiano Ronaldo'], 'height_cm']
```

```python
# get height and weight of Messi
df.loc['L. Messi', ['height_cm', 'weight_kg']]
```

```python
# get height and weight of Messi and Ronaldo
df.loc[['L. Messi', 'Cristiano Ronaldo'], ['height_cm', 'weight_kg']]
```

## Selecting a range of data with a slice

```python
# slice column labels
players = ['L. Messi', 'Cristiano Ronaldo']

df.loc[players, 'age':'club']
```

```python
# slice index labels
columns = ['age',	'dob', 'height_cm', 'weight_kg']

# get top1 and top10 player name
df.index[:10]

df.loc['L. Messi':'M. Salah', columns]
```

## Selecting with conditions

```python
# one: select player with height above 180cm
columns = ['age',	'dob', 'height_cm', 'weight_kg']


df.loc[df['height_cm']>180, columns]
```

```python
# multiple: select players with height above 180cm from Argentina
df.loc[(df['height_cm']>180) & (df['nationality']=="Argentina"), :]
```

## selecting a single value(iloc method)

```python
# height of messi
df.iloc[0, 3]
```

```python
# weight of ronaldo
df.iloc[1, 3]
```

```python
df
```

```python
# all rows inside 'height_cm' column
df.iloc[:, 3]
```

```python
# all rows that correspond to the index 'L. Messi'
df.iloc[0, :]
```

## selecting with a list of values

```python
# get all data about messi and ronaldo
df.iloc[[0,1]]
```

```python
# height of ronaldo and messi
df.iloc[[0,1], 3]
```

```python
# height and weight of messi
df.iloc[0, [3,4]]
```

```python
# height and weight of messi and ronaldo
df.iloc[[0,1], [3,4]]
```

## selecting a range of data with a slice

```python
# slice column labels age/ club
# players = ['L. Messi', 'Cristiano Ronaldo']
players = [0,1] 
df.iloc[players, 1:7]
```

```python
# slice index labels (top1 to top10 player)
# columns = ['age', 'dob', 'height_cm', 'weight_kg']
columns = [1,2,3,4]
df.iloc[0:10, columns]

# get top 1 and top 10 player names
```

## selecting with conditions

iloc only accepts boolean lists (not boolean series). have to use list() function to convert boolean series -> list

```python
# height above 180
columns = [1,2,3,4]
list(df['height_cm']>180)
df.iloc[list(df['height_cm']>180), columns]
```

```python
df.loc[(df['height_cm'] > 180) & (df['nationality'] == "Argentina")]
```

## Set value to one cell

```python
df
```

```python
# update messi height
df.loc['L. Messi', 'height_cm'] = 175
df
```

## Set value to entire column

```python
# set height of all players to 190
df.loc[:, 'height_cm'] = 190
df
```

## Set value to entire row

```python
# get all columns that correspond to player ranked last in fifa
df.iloc[-1, :]
```

```python
# set null values to all columns that correspond to player ranked last
import numpy as np 

df.iloc[-1, :] = np.nan
df
# all values in last row should be nan
```

## Set value for multiple cells

```python
# set value for all items matching the list of labels
df.loc[['L. Messi', 'Cristiano Ronaldo'], 'height_cm'] = 175
df
```

## Set value for rows matching a condition

```python
# change height only for certain players
columns = ['age', 'dob', 'height_cm', 'weight_kg']
df.loc[df['height_cm'] >180, columns] = 0
```
